## Step 39: Load the attendance dataset


```python
# Step 39: Load the attendance dataset
# read_csv() loads employee login and logout records.
attendance = pd.read_csv("attendance.csv")
```

## Step 40: Preview attendance data


```python
# Step 40: Preview attendance data
# Display the first 5 attendance records.
attendance.head()
```

## Step 41: Check the size of the attendance dataset


```python
# Step 41: Check the size of the attendance dataset
# shape returns (number of rows, number of columns).
attendance.shape
```

## Step 42: Check for missing attendance values


```python
# Step 42: Check for missing attendance values
# Missing Login or Logout values can affect working-hour calculations.
attendance.isnull().sum()
```

## Step 43: Convert Login and Logout into datetime values


```python
# Step 43: Convert Login and Logout into datetime values
# to_datetime() converts text such as 09:30 into a time value.
attendance["Login"] = pd.to_datetime(
    attendance["Login"],
    format="%H:%M"
)

attendance["Logout"] = pd.to_datetime(
    attendance["Logout"],
    format="%H:%M"
)
```

## Step 44: Calculate working hours


```python
# Step 44: Calculate working hours
# Subtract Login from Logout to get a time duration.
# total_seconds() converts the duration to seconds.
# Dividing by 3600 converts seconds into hours.
attendance["WorkingHours"] = (
    attendance["Logout"] - attendance["Login"]
).dt.total_seconds() / 3600
```

## Step 45: Preview attendance with working hours


```python
# Step 45: Preview attendance with working hours
# Check the new WorkingHours column.
attendance.head()
```

## Step 46: Inspect the raw Logout - Login time difference


```python
# Step 46: Inspect the raw Logout - Login time difference
# The result is shown as a time duration (timedelta).
attendance["Logout"] - attendance["Login"]
```

## Step 47: Recalculate working hours


```python
# Step 47: Recalculate working hours
# This repeats the working-hour calculation as a verification step.
attendance["WorkingHours"] = (
    attendance["Logout"] - attendance["Login"]
).dt.total_seconds() / 3600
```

## Step 48: Recheck the attendance data


```python
# Step 48: Recheck the attendance data
# Display the first 5 rows after the working-hour calculation.
attendance.head()
```

## Step 49: Find invalid attendance records


```python
# Step 49: Find invalid attendance records
# A negative WorkingHours value means Logout is earlier than Login
# in the same-day calculation, so the record is treated as an anomaly.
invalid_attendance = attendance[
    attendance["WorkingHours"] < 0
]

invalid_attendance
```

## Step 50: Create a clean attendance dataset


```python
# Step 50: Create a clean attendance dataset
# Keep only records with non-negative WorkingHours.
attendance_clean = attendance[
    attendance["WorkingHours"] >= 0
].copy()
```

## Step 51: Check the minimum valid working hours


```python
# Step 51: Check the minimum valid working hours
# This confirms that the cleaned data no longer contains negative values.
attendance_clean["WorkingHours"].min()
```

## Step 52: Calculate average working hours for each employee


```python
# Step 52: Calculate average working hours for each employee
attendance_summary = (
    attendance_clean
    .groupby("EmployeeID")["WorkingHours"]
    .mean()
    .sort_values()
)

# Display the employee-wise average working hours.
attendance_summary
```

## Step 53: Display the 10 employees with the lowest average working hours


```python
# Step 53: Display the 10 employees with the lowest average working hours
# The summary is sorted from lowest to highest.
attendance_summary.head(10)
```

## Step 54: Convert the attendance summary into a DataFrame


```python
# Step 54: Convert the attendance summary into a DataFrame
attendance_summary_df = attendance_summary.reset_index()

# Rename the columns for clarity.
attendance_summary_df.columns = [
    "EmployeeID",
    "AverageWorkingHours"
]

attendance_summary_df.head()
```

## Step 55: Display the complete attendance summary


```python
# Step 55: Display the complete attendance summary
attendance_summary
```

## Step 56: Count attendance anomalies by employee


```python
# Step 56: Count attendance anomalies by employee
# Keep negative working-hour records, group them by EmployeeID,
# and count how many anomalies each employee has.
attendance_anomaly_summary = (
    attendance[attendance["WorkingHours"] < 0]
    .groupby("EmployeeID")
    .size()
    .reset_index(name="AttendanceAnomalies")
)

attendance_anomaly_summary
```

## Step 57: Count attendance anomalies for every employee


```python
# Step 57: Count attendance anomalies for every employee
# Create a True/False Anomaly column.
# sum() counts True values because True behaves like 1 and False like 0.
employee_attendance_anomalies = (
    attendance.assign(
        Anomaly=attendance["WorkingHours"] < 0
    )
    .groupby("EmployeeID")["Anomaly"]
    .sum()
    .reset_index(name="AttendanceAnomalies")
)

employee_attendance_anomalies
```

## Step 58: Review the current employee evidence table


```python
# Step 58: Review the current employee evidence table
# It currently contains transaction totals and refund totals.
employee_evidence
```

## Step 59: Add attendance anomaly counts to employee evidence


```python
# Step 59: Add attendance anomaly counts to employee evidence
# Merge the attendance anomaly information using EmployeeID.
employee_evidence = employee_evidence.merge(
    employee_attendance_anomalies,
    on="EmployeeID",
    how="left"
)
```

## Step 60: Preview the evidence table after the attendance merge


```python
# Step 60: Preview the evidence table after the attendance merge
employee_evidence.head()
```

## Step 61: Replace missing anomaly counts with zero


```python
# Step 61: Replace missing anomaly counts with zero
# Employees without anomalies receive NaN after the merge.
# fillna(0) changes them to 0, and astype(int) makes the count an integer.
employee_evidence["AttendanceAnomalies"] = (
    employee_evidence["AttendanceAnomalies"]
    .fillna(0)
    .astype(int)
)
```

## Step 62: Rank employees by total transaction amount


```python
# Step 62: Rank employees by total transaction amount
employee_evidence.sort_values(
    by="TotalTransactionAmount",
    ascending=False
).head(10)
```

## Step 63: Rank employees by total refund amount


```python
# Step 63: Rank employees by total refund amount
employee_evidence.sort_values(
    by="TotalRefundAmount",
    ascending=False
).head(10)
```

## Step 64: Rank employees by attendance anomalies


```python
# Step 64: Rank employees by attendance anomalies
employee_evidence.sort_values(
    by="AttendanceAnomalies",
    ascending=False
).head(10)
```

## Step 65: Recheck attendance-anomaly ranking


```python
# Step 65: Recheck attendance-anomaly ranking
# This repeats the ranking before building the final risk indicators.
employee_evidence.sort_values(
    by="AttendanceAnomalies",
    ascending=False
).head(10)
```

## Step 66: Recheck refund ranking


```python
# Step 66: Recheck refund ranking
employee_evidence.sort_values(
    by="TotalRefundAmount",
    ascending=False
).head(10)
```

## Step 67: Recheck attendance ranking


```python
# Step 67: Recheck attendance ranking
employee_evidence.sort_values(
    by="AttendanceAnomalies",
    ascending=False
).head(10)
```

## Step 68: Calculate the refund rate


```python
# Step 68: Calculate the refund rate
# RefundRate = total refunds / total transactions × 100.
employee_evidence["RefundRate"] = (
    employee_evidence["TotalRefundAmount"]
    / employee_evidence["TotalTransactionAmount"]
) * 100
```

## Step 69: Rank employees by refund rate


```python
# Step 69: Rank employees by refund rate
# Higher refund rates appear first.
employee_evidence.sort_values(
    by="RefundRate",
    ascending=False
).head(10)
```

## Step 70: Calculate 90th-percentile thresholds


```python
# Step 70: Calculate 90th-percentile thresholds
# These thresholds identify unusually high values for each risk indicator.
transaction_threshold = employee_evidence[
    "TotalTransactionAmount"
].quantile(0.90)

refund_threshold = employee_evidence[
    "TotalRefundAmount"
].quantile(0.90)

refund_rate_threshold = employee_evidence[
    "RefundRate"
].quantile(0.90)

print("Transaction threshold:", transaction_threshold)
print("Refund threshold:", refund_threshold)
print("Refund rate threshold:", refund_rate_threshold)
```

## Step 71: Flag unusually high transaction totals


```python
# Step 71: Flag unusually high transaction totals
# True means the employee is at or above the transaction threshold.
employee_evidence["HighTransaction"] = (
    employee_evidence["TotalTransactionAmount"]
    >= transaction_threshold
)
```

## Step 72: Flag unusually high refund totals


```python
# Step 72: Flag unusually high refund totals
# True means the employee is at or above the refund threshold.
employee_evidence["HighRefund"] = (
    employee_evidence["TotalRefundAmount"]
    >= refund_threshold
)
```

## Step 73: Flag unusually high refund rates


```python
# Step 73: Flag unusually high refund rates
# True means the employee is at or above the refund-rate threshold.
employee_evidence["HighRefundRate"] = (
    employee_evidence["RefundRate"]
    >= refund_rate_threshold
)
```

## Step 74: Flag attendance issues


```python
# Step 74: Flag attendance issues
# True means the employee has at least one attendance anomaly.
employee_evidence["AttendanceIssue"] = (
    employee_evidence["AttendanceAnomalies"] > 0
)
```

## Step 75: Calculate the Investigation Score


```python
# Step 75: Calculate the Investigation Score
# Each True flag becomes 1 and each False flag becomes 0.
# The final score can range from 0 to 4.
employee_evidence["InvestigationScore"] = (
    employee_evidence["HighTransaction"].astype(int)
    + employee_evidence["HighRefund"].astype(int)
    + employee_evidence["HighRefundRate"].astype(int)
    + employee_evidence["AttendanceIssue"].astype(int)
)
```

## Step 76: Create the final investigation ranking


```python
# Step 76: Create the final investigation ranking
# Sort employees from the highest InvestigationScore to the lowest.
final_ranking = employee_evidence.sort_values(
    by="InvestigationScore",
    ascending=False
)

# Display the key evidence for the top 10 candidates.
final_ranking[
    [
        "EmployeeID",
        "TotalTransactionAmount",
        "TotalRefundAmount",
        "RefundRate",
        "AttendanceAnomalies",
        "InvestigationScore"
    ]
].head(10)
```

## Step 77: Display the top investigation candidate


```python
# Step 77: Display the top investigation candidate
# iloc[0] selects the first row of the final ranking.
top_candidate = final_ranking.iloc[0]

print("========================================")
print("       🕵️ DATA DETECTIVE RESULT")
print("========================================")
print("Investigation Candidate:", top_candidate["EmployeeID"])
print("Transaction Amount: ₹", top_candidate["TotalTransactionAmount"])
print("Refund Amount: ₹", top_candidate["TotalRefundAmount"])
print("Refund Rate:", round(top_candidate["RefundRate"], 2), "%")
print("Attendance Anomalies:", top_candidate["AttendanceAnomalies"])
print("Investigation Score:", top_candidate["InvestigationScore"], "/ 4")
print("========================================")
print("⚠️ FLAGGED FOR FURTHER INVESTIGATION")
print("========================================")
```

## Step 78: Recalculate and recheck the refund rate


```python
# Step 78: Recalculate and recheck the refund rate
# This is a final verification of the refund-rate calculation.
employee_evidence["RefundRate"] = (
    employee_evidence["TotalRefundAmount"]
    / employee_evidence["TotalTransactionAmount"]
) * 100

employee_evidence.sort_values(
    by="RefundRate",
    ascending=False
).head(10)
```

```python
# Step 81: Create the interactive 3D investigation map
# X-axis = total transactions
# Y-axis = total refunds
# Z-axis = refund rate
# Color and size = InvestigationScore
fig = px.scatter_3d(
    employee_evidence,
    x="TotalTransactionAmount",
    y="TotalRefundAmount",
    z="RefundRate",
    color="InvestigationScore",
    size="InvestigationScore",
    hover_name="EmployeeID",
    hover_data=[
        "TotalTransactionAmount",
        "TotalRefundAmount",
        "RefundRate",
        "AttendanceAnomalies"
    ],
    title="🕵️ Data Detective — 3D Investigation Map",
    color_continuous_scale="Turbo"
)

# Add clear axis titles and set the chart height.
fig.update_layout(
    scene=dict(
        xaxis_title="Total Transactions",
        yaxis_title="Total Refunds",
        zaxis_title="Refund Rate (%)"
    ),
    height=750
)

# Display the interactive 3D chart.
fig.show()
```
